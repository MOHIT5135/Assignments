1. Employees earning more than average salary


SELECT name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees);

Explanation: 
* The subquery (SELECT AVG(salary) FROM employees) calculates average salary.

* Outer query compares each employee’s salary with this value.

* Only employees with higher-than-average salaries are returned.


2. Top 5 highest-paid employees

SELECT name, salary
FROM employees
ORDER BY salary DESC
LIMIT 5;

Explanation:
* ORDER BY salary DESC sorts salaries from highest to lowest.

* LIMIT 5 picks the top 5 records.

* Sorting is crucial—without it, you’d get random 5 employees.

3. Aggregate calculations


SELECT 
    COUNT(*) AS total_employees,
    AVG(salary) AS avg_salary,
    MIN(salary) AS min_salary,
    MAX(salary) AS max_salary
FROM employees;

Explanation:

* Aggregate functions: work on multiple rows (COUNT, AVG, MIN, MAX).

* Scalar functions: work on single values (e.g., UPPER(), ROUND()).

4. Total sales per region (with filter)

SELECT region, SUM(amount) AS total_sales
FROM sales
GROUP BY region
HAVING SUM(amount) > 50000;


Explanation:

* GROUP BY region groups data.

* SUM(amount) calculates total per region.

* HAVING filters grouped results (not WHERE).

5. Number of unique job roles


SELECT COUNT(DISTINCT job_role) AS unique_roles
FROM employees;

Explanation:

* DISTINCT removes duplicates.

* Without it, COUNT would include repeated roles.


6. Students with marks between 60 and 80

Normal condition:
SELECT *
FROM students
WHERE marks >= 60 AND marks <= 80;

Using BETWEEN:
SELECT *
FROM students
WHERE marks BETWEEN 60 AND 80;


Explanation:

* BETWEEN is inclusive (includes 60 and 80).

7. Employees with NULL commission

SELECT *
FROM employees
WHERE commission IS NULL;

Explanation:

* Use IS NULL (NOT = NULL).

* SQL treats NULL as unknown, so = doesn’t work.

8. Increase salary by 10% (IT department)

UPDATE employees
SET salary = salary * 1.10
WHERE department = 'IT';

Explanation:

* Arithmetic operations are applied row by row.

* Salary * 1.10 increases salary by 10%.

9. Delete students with marks < 40

DELETE FROM students
WHERE marks < 40;



10. Employees earning more than department average (no joins)

SELECT name, salary, department
FROM employees e
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
    WHERE department = e.department
);


Explanation:

* Subquery calculates average salary per department.

* e.department links outer and inner query (correlated subquery).

* Each employee is compared with their own department’s average.
